```python
# Bibliotecas
import os
from dotenv import load_dotenv
from langchain_community.document_loaders import PyPDFDirectoryLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
```

```python
# Carrega as variáveis de ambiente do arquivo .env
load_dotenv()

# Definição dos caminhos das pastas
DATA_PATH = "../data/raw_contracts"
DB_PATH = "../data/vectorstore"

print("Bibliotecas importadas e caminhos configurados")
```

```python
# Inicializa o loader apontando para a pasta com os 5 PDFs de teste
loader = PyPDFDirectoryLoader(DATA_PATH)
raw_documents = loader.load()

print(f"Total de páginas carregadas: {len(raw_documents)}")

# Exibe um techo do primeiro documento para validação visual
print("Amostra do texto")
print(raw_documents[0].page_content[:300])
```

```python
# Configuração da estratégia de divisão
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000, # Tamanho máximo de cada bloco de texto
    chunk_overlap=200, # Sobreposição para não perder o contexto entre blocos
    length_function=len
)

# Aplica a divisão aos documentos brutos
chunks = text_splitter.split_documents(raw_documents)

print(f"Os documentos foram divididos em {len(chunks)} blocos (chunks).")
print("Amostra de um chunk")
print(chunks[0].page_content)
```

```python
# Inicializa o modelo open-source gratuito (não precisa de API Key)
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# Cria e persiste o banco de dados vetorial localmente no diretório definido
vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    persist_directory=DB_PATH
)

print(f"Vetorização concluída com sucesso! Banco salvo em: {DB_PATH}")
```
